```python
x=df.drop(columns=['price'])
y=df['price']
```

```python
x
```

# Preprocessing

```python
x_one_enc = pd.get_dummies(x,columns=['model','transmission','fuelType'],drop_first=True)
```

```python
x_one_enc
```

```python
x_one_enc=x_one_enc.astype(int)
```

```python
x_one_enc
```

```python
from sklearn.preprocessing import LabelEncoder

encode = LabelEncoder()

column = ['model','transmission','fuelType']

Xlable = x

for i in column:
   Xlable[i]= encode.fit_transform(Xlable[i])
```

```python
Xlable
```

```python
from sklearn.preprocessing import StandardScaler

numerical_cols = ['year','mileage','tax','mpg','engineSize']
scaler = StandardScaler()
x_one_enc[numerical_cols] = scaler.fit_transform(x_one_enc[numerical_cols])
```

```python
x_one_enc
```

```python
Xlable.columns
```

```python
Xlable[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg','engineSize']] = scaler.fit_transform(Xlable[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg','engineSize']])
```

```python
Xlable
```

# Modeling

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
X_train, X_test, y_train, y_test = train_test_split(x_one_enc, y, test_size=0.33, random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
r2 = r2_score(y_test,y_pred)
r2
```

```python
n = X_test.shape[0]
p = X_test.shape[1]
adjusted_r2 = 1-((1 - r2) * (n - 1))/(n - p - 1)
adjusted_r2
```

```python
X_train, X_test, y_train, y_test = train_test_split(Xlable, y, test_size=0.33, random_state=42)
```

```python
model2 = LinearRegression()
model2.fit(X_train,y_train)
```

```python
y_pred = model2.predict(X_test)
y_pred
```

```python
r2 = r2_score(y_test,y_pred)
r2
```

```python
n = X_test.shape[0]
p = X_test.shape[1]
adjusted_r2 = 1-((1 - r2) * (n - 1))/(n - p - 1)
adjusted_r2
```
